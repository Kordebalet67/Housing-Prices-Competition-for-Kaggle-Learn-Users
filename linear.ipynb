```python
df = pd.read_csv('processed_dataset.csv')
df
```

```python
df = df.fillna(0)
```

```python
for column in df.columns:
    print(column)
    print(df[column].unique())
```

```python
# Фильтрация датафрейма по назначению данных
train_df = df.loc[df['set_type'] == 'train']
val_df = df.loc[df['set_type'] == 'val']
test_df = df.loc[df['set_type'] == 'test']

train_df = train_df.drop('set_type', axis=1)
val_df = val_df.drop('set_type', axis=1)
test_df = test_df.drop('set_type', axis=1)
test_df = test_df.drop('sale_price', axis=1)
```

```python
y_train = train_df['sale_price']
y_val = val_df['sale_price']
X_train = train_df.drop('sale_price', axis=1)
X_val = val_df.drop('sale_price', axis=1)
```

регрессии ругаются на NaN. придётся менять на 0

```python
degree = 2
poly_features = PolynomialFeatures(degree=degree)

X_poly_train = poly_features.fit_transform(X_train)
X_poly_val = poly_features.fit_transform(X_val)
X_poly_test = poly_features.fit_transform(test_df)
```

```python
poly_regressor = LinearRegression()
poly_regressor.fit(X_poly_train, y_train)

y_val_pred = poly_regressor.predict(X_poly_val)
y_train_pred = poly_regressor.predict(X_poly_train)
```

```python
ridge_regressor = Ridge(alpha=5.294024400913717)
lasso_regressor = Lasso()
```

```python
ridge_regressor.fit(X_poly_train, y_train)
lasso_regressor.fit(X_poly_train, y_train)
```

```python
y_pred_train_ridge = ridge_regressor.predict(X_poly_train)
y_pred_train_lasso = lasso_regressor.predict(X_poly_train)

y_pred_val_ridge = ridge_regressor.predict(X_poly_val)
y_pred_val_lasso = lasso_regressor.predict(X_poly_val)
```

```python
def objective(trial):
    # Параметры, которые мы хотим оптимизировать
    alpha = trial.suggest_loguniform('alpha', 1e-8, 100.0)

    # Инициализация модели
    model = Lasso(alpha=alpha)

    # Оценка модели с помощью кросс-валидации
    score = np.mean(cross_val_score(model, X_poly_train, y_train, scoring='neg_mean_squared_error', cv=3))

    return -score  # Поскольку Optuna максимизирует целевую функцию, умножаем на -1

# Создание и запуск оптимизации Optuna
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=100)

# Получение лучших параметров
best_params = study.best_params

# Печать лучших параметров
print("Лучшие параметры:", best_params)
```

```python
poly_train_mse = MSE(y_train, y_train_pred)
poly_train_mae = MAE(y_train, y_train_pred)

poly_val_mse = MSE(y_val, y_val_pred)
poly_val_mae = MAE(y_val, y_val_pred)

print(f"Polynomial Regression\n"
      f"\ttrain MSE = {poly_train_mse:.2f}, MAE = {poly_train_mae:.4f}\n"
      f"\tval MSE = {poly_val_mse:.2f}, MAE = {poly_val_mae:.4f}")
```

```python
train_mse_ridge = MSE(y_train, y_pred_train_ridge)
train_mae_ridge = MAE(y_train, y_pred_train_ridge)

val_mse_ridge = MSE(y_val, y_pred_val_ridge)
val_mae_ridge = MAE(y_val, y_pred_val_ridge)

print(f"Ridge\n"
      f"\ttrain MSE = {train_mse_ridge:.2f}, MAE = {train_mae_ridge:.2f}\n"
      f"\tval MSE = {val_mse_ridge:.2f}, MAE = {val_mae_ridge:.2f}")
```
